## Programación Dinámica. Viaje por el rio
* **Definición**: Es posible dividir el problema en subproblemas más pequeños, guardando las soluciones para ser utilizadas más adelante.
* **Características** que permiten identificar problemas aplicables:<br>
  -Es posible almacenar soluciones de los subproblemas para ser utilizados más adelante<br>
  -Debe verificar el principio de optimalidad de Bellman: “en una secuencia optima de decisiones, toda sub-secuencia también es óptima” (*)<br>
  -La necesidad de guardar la información acerca de las soluciones parciales unido a la recursividad provoca la necesidad de preocuparnos por la complejidad espacial (cuantos recursos de espacio usaremos)<br>

### Problema<br>
En un río hay **n** embarcaderos y debemos desplazarnos río abajo desde un embarcadero a otro. Cada embarcadero tiene precios diferentes para ir de un embarcadero a otro situado más abajo. Para ir del embarcadero i al j, puede ocurrir que sea más barato hacer un trasbordo por un embarcadero intermedio k. El problema consiste en determinar la combinación más barata.

1.jpg

*Consideramos una tabla TARIFAS(i,j) para almacenar todos los precios que nos ofrecen los embarcaderos.<br>
*Si no es posible ir desde i a j daremos un valor alto para garantizar que ese trayecto no se va a elegir en la ruta óptima(modelado habitual para restricciones)

```python
#Viaje por el rio - Programación dinámica
################################################################################

TARIFAS = [
[0,5,4,3,float("inf"),999,999],   #desde nodo 0
[999,0,999,2,3,999,11],  #desde nodo 1
[999,999, 0,1,999,4,10], #desde nodo 2
[999,999,999, 0,5,6,9],
[999,999, 999,999,0,999,4],
[999,999, 999,999,999,0,3],
[999,999,999,999,999,999,0]
]

#999 se puede sustituir por float("inf") del modulo math
TARIFAS
```

```python
#Calculo de la matriz de PRECIOS y RUTAS
#  PRECIOS - contiene la matriz del mejor precio para ir de un nodo a otro
#  RUTAS   - contiene los nodos intermedios para ir de un nodo a otro
################################################################
def Precios(TARIFAS):
################################################################
  #Total de Nodos
  N = len(TARIFAS[0])

  #Inicialización de la tabla de precios
  PRECIOS = [ [9999]*N for i in [9999]*N]  #n x n
  RUTA = [ [""]*N for i in [""]*N]

  #Se recorren todos los nodos con dos bucles(origen - destino)
  #  para ir construyendo la matriz de PRECIOS
  for i in range(N-1):
    for j in range(i+1, N):

      # Asumimos que el mínimo es la tarifa de un nodo a otro inicialmente
      MIN = TARIFAS[i][j]
      print("La tarifa desde " + str(i) + " hasta " + str(j) + " con un precio de " + str(TARIFAS[i][j]))

      # Asumimos que la mejor ruta de un nodo a otro comienza en el nodo inicial (ej: 0,1 -> 0, 0,4 -> 0, 2,5 -> 2)
      RUTA[i][j] = i

      # Recorremos posibles movimientos intermedios
      for k in range(i, j):

         # Si el precio de ir de i a k más la tarifa de k a j es menor que la tarifa de i a j, nos quedamos con esa ruta
        if PRECIOS[i][k] + TARIFAS[k][j] < MIN:
            MIN = min(MIN, PRECIOS[i][k] + TARIFAS[k][j] )
            RUTA[i][j] = k
            print("Ruta indirecta identificada desde " + str(i) + " hasta " + str(j) + " pasando por " + str(k) + " con un precio de " + str(MIN))

        # Actualizamos precio de i a j como el mínimo (puede ser pasando por k)
        PRECIOS[i][j] = MIN

  return PRECIOS,RUTA
```

```python
PRECIOS,RUTA = Precios(TARIFAS)
#print(PRECIOS[0][6])

print("PRECIOS")
for i in range(len(TARIFAS)):
  print(PRECIOS[i])

print("\nRUTA")
for i in range(len(TARIFAS)):
  print(RUTA[i])
```

```python
#Calculo de la ruta usando la matriz RUTA
def calcular_ruta(RUTA, desde, hasta):
  if desde == RUTA[desde][hasta]:
  #if desde == hasta:
    #print("Ir a :" + str(desde))
    return desde
  else:
    return str(calcular_ruta(RUTA, desde, RUTA[desde][hasta])) +  ',' + str(RUTA[desde][hasta])

print("\nLa ruta es:")
calcular_ruta(RUTA, 0,6)
```

## Problema de Asignacion de tarea

```python
#Asignacion de tareas - Ramificación y Poda
################################################################################
#    T A R E A
#   A
#   G
#   E
#   N
#   T
#   E

COSTES=[[11,12,18,40],
        [14,15,13,22],
        [11,17,19,23],
        [17,14,20,28]]
```

```python
#Calculo del valor de una solucion parcial
def valor(S,COSTES):
  VALOR = 0
  for i in range(len(S)):
    VALOR += COSTES[S[i]][i]
  return VALOR



valor((3,2,  ),COSTES)
```

```python
#Coste inferior para soluciones parciales
#  (1,3,) Se asigna la tarea 1 al agente 0 y la tarea 3 al agente 1

def CI(S,COSTES):
    # Calculo el coste parcial de la solución dada
    coste = valor(S,COSTES)

    # Calculo coste mínimo de opciones restantes no usadas permitiendo reutilización por sencillez (relajación = menos poda)
    no_usadas = set(range(len(COSTES)))-set(S)

    for i in range(len(S),len(COSTES)):
        coste += min([COSTES[j][i] for j in no_usadas])

    return coste

def CS(S,COSTES):
    # Calculo el coste parcial de la solución dada
    coste = valor(S,COSTES)

    # Calculo coste máximo de opciones restantes no usadas
    no_usadas = set(range(len(COSTES)))-set(S)

    for i in range(len(S),len(COSTES)):
      coste += max([COSTES[j][i] for j in no_usadas])

    return coste

CI((0,1),COSTES)
```

```python
#Genera tantos hijos como como posibilidades haya para la siguiente elemento de la tupla
#(0,) -> (0,1), (0,2), (0,3)
def crear_hijos(NODO, N):
  HIJOS = []
  for i in range(N ):
    if i not in NODO:
      HIJOS.append({'s':NODO +(i,)    })
  return HIJOS
```

```python
def ramificacion_y_poda(COSTES):
#Construccion iterativa de soluciones(arbol). En cada etapa asignamos un agente(ramas).
#Nodos del grafo  { s:(1,2),CI:3,CS:5  }
  print(COSTES)
  DIMENSION = len(COSTES)
  MEJOR_SOLUCION=tuple( i for i in range(len(COSTES)) )
  CotaSup = valor(MEJOR_SOLUCION,COSTES)
  print("Cota Superior:", CotaSup)

  NODOS=[]
  NODOS.append({'s':(), 'ci':CI((),COSTES)    } )

  iteracion = 0

  while( len(NODOS) > 0):
    iteracion +=1

    nodo_prometedor = [ min(NODOS, key=lambda x:x['ci']) ][0]['s']
    print("Nodo prometedor:", nodo_prometedor)

    #Ramificacion
    #Se generan los hijos
    HIJOS =[ {'s':x['s'], 'ci':CI(x['s'], COSTES)   } for x in crear_hijos(nodo_prometedor, DIMENSION) ]
    print("Explorando hijos:")
    print(HIJOS)

    #Revisamos la cota superior y nos quedamos con la mejor solucion si llegamos a una solucion final
    NODO_FINAL = [x for x in HIJOS if len(x['s']) == DIMENSION  ]
    if len(NODO_FINAL ) >0:
      print("\n********Soluciones:",  [x for x in HIJOS if len(x['s']) == DIMENSION  ] )
      if NODO_FINAL[0]['ci'] < CotaSup:
        CotaSup = NODO_FINAL[0]['ci']
        MEJOR_SOLUCION = NODO_FINAL

    #Poda
    HIJOS = [x for x in HIJOS if x['ci'] < CotaSup   ]
    print("Despues de la poda:")
    print(HIJOS)

    #Añadimos los hijos
    NODOS.extend(HIJOS)

    #Eliminamos el nodo ramificado
    NODOS =  [  x for x in NODOS if x['s'] != nodo_prometedor    ]
    print("NODOS a explorar: ")
    print(NODOS)

  print("La solucion final es:" ,MEJOR_SOLUCION , " en " , iteracion , " iteraciones" , " para dimension: " ,DIMENSION  )



ramificacion_y_poda(COSTES)
```

## Fuerza Bruta (actividad extra)

```python
import time
import itertools

def fuerza_bruta(costes):
  #Representacion de la solucion: una tupla donde cada valor en la coordenada i-esima es la tarea asignada al agente i
  #Cuantas posibilidades hay? n! -> complejidad factorial (exponencial)

  mejor_valor = float('inf')
  mejor_solucion = ()

  for s in list(itertools.permutations(range(len(costes)))):
    valor_tmp = valor(s,costes)
    if valor_tmp < mejor_valor:
        mejor_valor = valor_tmp
        mejor_solucion = s

  print("La mejor solucion es :", mejor_solucion, " con valor:", mejor_valor)
  return mejor_solucion, mejor_valor

inicio = time.perf_counter()
fuerza_bruta(COSTES)
print("Tiempo:", time.perf_counter() - inicio, "s")
```

## Análisis para mejorar nota

Fuerza bruta prueba las n! permutaciones posibles (O(n!)). Ramificación y poda también es exponencial en el peor caso, pero poda ramas y en la práctica es mucho más rápida.

Se generan matrices aleatorias de tamaño creciente y se mide el tiempo real de cada algoritmo.

```python
import random

def generar_costes(n, minimo=1, maximo=100):
  return [[random.randint(minimo, maximo) for _ in range(n)] for _ in range(n)]


def fuerza_bruta_sin_print(costes):
  mejor_valor = float('inf')
  mejor_solucion = ()
  for s in itertools.permutations(range(len(costes))):
    valor_tmp = valor(s,costes)
    if valor_tmp < mejor_valor:
      mejor_valor = valor_tmp
      mejor_solucion = s
  return mejor_solucion, mejor_valor


def ramificacion_y_poda_sin_print(costes):
  dimension = len(costes)
  mejor_solucion = tuple(i for i in range(dimension))
  cota_sup = valor(mejor_solucion, costes)
  nodos = [{'s': (), 'ci': CI((), costes)}]
  while len(nodos) > 0:
    nodo_prometedor = min(nodos, key=lambda x: x['ci'])['s']
    hijos = [{'s': x['s'], 'ci': CI(x['s'], costes)} for x in crear_hijos(nodo_prometedor, dimension)]
    nodo_final = [x for x in hijos if len(x['s']) == dimension]
    if nodo_final and nodo_final[0]['ci'] < cota_sup:
      cota_sup = nodo_final[0]['ci']
      mejor_solucion = nodo_final[0]['s']
    hijos = [x for x in hijos if x['ci'] < cota_sup]
    nodos.extend(hijos)
    nodos = [x for x in nodos if x['s'] != nodo_prometedor]
  return mejor_solucion, cota_sup
```

```python
tamanios_fuerza_bruta = [4, 6, 8, 9, 10]
tiempos_fuerza_bruta = []

print(f"{'n':>4} {'tiempo (s)':>12}")
for n in tamanios_fuerza_bruta:
  costes = generar_costes(n)
  inicio = time.perf_counter()
  fuerza_bruta_sin_print(costes)
  tiempo = time.perf_counter() - inicio
  tiempos_fuerza_bruta.append(tiempo)
  print(f"{n:>4} {tiempo:>12.4f}")
```

```python
tamanios_ramificacion = [4, 6, 8, 10, 12, 14, 16]
tiempos_ramificacion = []

print(f"{'n':>4} {'tiempo (s)':>12}")
for n in tamanios_ramificacion:
  costes = generar_costes(n)
  inicio = time.perf_counter()
  ramificacion_y_poda_sin_print(costes)
  tiempo = time.perf_counter() - inicio
  tiempos_ramificacion.append(tiempo)
  print(f"{n:>4} {tiempo:>12.4f}")
```

```python
import matplotlib.pyplot as plt

plt.plot(tamanios_fuerza_bruta, tiempos_fuerza_bruta, marker='o', label='Fuerza bruta')
plt.plot(tamanios_ramificacion, tiempos_ramificacion, marker='o', label='Ramificación y poda')
plt.xlabel('n')
plt.ylabel('tiempo (s)')
plt.yscale('log')
plt.legend()
plt.show()
```

### Conclusión

Fuerza bruta crece muy rápido (factorial): en un par de incrementos de n pasa de ser prácticamente instantánea a tardar más de un segundo, y deja de ser viable a partir de n≈12-13.

Ramificación y poda también es exponencial en el peor caso, así que el tiempo también se dispara al crecer n, pero al podar ramas es consistentemente mucho más rápida que fuerza bruta para el mismo tamaño de problema. Los tiempos exactos varían según la máquina y los datos aleatorios de cada ejecución, pero el patrón se mantiene.

Vamos a buscar el minimo de la funcion paraboloide : $$f(x) =  x² + y²$$

```python
#Definimos la funcion
#Paraboloide
f  = lambda X:      X[0]**2 + X[1]**2    #Funcion
df = lambda X: [2*X[0] , 2*X[1]]         #Gradiente

df([1,2])
```

```python
#Definimos la funcion
f= lambda X: math.sin(1/2 * X[0]**2 - 1/4 * X[1]**2 + 3) *math.cos(2*X[0] + 1 - math.exp(X[1]) )
```

## Práctica para mejorar nota (actividad extra)

Minimizar f(x,y) por descenso del gradiente. No hay derivada dada, así que se deriva a mano aplicando la regla de la cadena (mismo procedimiento que con la función anterior, x²+y²), reutilizando el mismo bucle de descenso de la sección anterior.

También se reutilizan la misma región del mapa de niveles (resolución y rango) y la misma zona para generar el punto aleatorio inicial que en el ejemplo anterior, ya que es el mismo escenario, solo cambia la función a minimizar.

```python
def df(X):
  u = 1/2 * X[0]**2 - 1/4 * X[1]**2 + 3
  v = 2*X[0] + 1 - math.exp(X[1])
  dfdx = X[0]*math.cos(u)*math.cos(v) - 2*math.sin(u)*math.sin(v)
  dfdy = -X[1]/2*math.cos(u)*math.cos(v) + math.exp(X[1])*math.sin(u)*math.sin(v)
  return [dfdx, dfdy]


def limitar(v, minimo=-5, maximo=5):
  return max(min(v, maximo), minimo)
```

```python
#Prepara los datos para dibujar mapa de niveles de Z
resolucion = 100
rango = 5.5

x_vals = np.linspace(-rango, rango, resolucion)
y_vals = np.linspace(-rango, rango, resolucion)
z = np.zeros((resolucion, resolucion))
for ix, x in enumerate(x_vals):
  for iy, y in enumerate(y_vals):
    z[iy, ix] = f([x, y])

#Pinta el mapa de niveles de Z
plt.contourf(x_vals, y_vals, z, resolucion)
plt.colorbar()

#Punto aleatorio inicial, pintado de blanco
p = [random.uniform(-5, 5), random.uniform(-5, 5)]
plt.plot(p[0], p[1], "o", c="white")

tasa_aprendizaje = .1

for _ in range(50):
  grad = df(p)
  p[0] = limitar(p[0] - tasa_aprendizaje*grad[0])
  p[1] = limitar(p[1] - tasa_aprendizaje*grad[1])
  plt.plot(p[0], p[1], "o", c="red")

#Punto final, pintado de verde
plt.plot(p[0], p[1], "o", c="green")
plt.show()
print("Solucion:", p, f(p))
```

La función tiene varios mínimos locales (se ve en el mapa de niveles de arriba y en el gráfico 3D de la sección anterior). Un solo punto de partida no garantiza el mínimo global, así que se repite el descenso desde varios puntos aleatorios y se guarda el mejor resultado.

```python
mejor_p = None
mejor_valor = float('inf')

for _ in range(30):
  p = [random.uniform(-5, 5), random.uniform(-5, 5)]
  for _ in range(100):
    grad = df(p)
    p[0] = limitar(p[0] - tasa_aprendizaje*grad[0])
    p[1] = limitar(p[1] - tasa_aprendizaje*grad[1])
  valor_actual = f(p)
  if valor_actual < mejor_valor:
    mejor_valor = valor_actual
    mejor_p = p

print("Mejor solucion encontrada:", mejor_p, mejor_valor)
```
