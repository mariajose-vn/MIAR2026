## Devolución de cambio por técnica voraz

```python
def cambio_monedas(N, SM):
  SOLUCION = [0]*len(SM)   #SOLUCION = [0,0,0,0,..]
  ValorAcumulado = 0

  for i,valor in enumerate(SM):
    monedas =  (N-ValorAcumulado)//valor
    SOLUCION[i] = monedas
    ValorAcumulado = ValorAcumulado + monedas*valor

    if ValorAcumulado == N:
      return SOLUCION


cambio_monedas(15,[25,10,5,1])
```

## Práctica Individual - Encontrar los dos puntos más cercanos

Problema: dado un conjunto de puntos, encontrar los dos más cercanos entre sí (distancia euclídea).

Guía:
1. 1D: fuerza bruta → complejidad → ¿se puede mejorar?
2. 1D: Divide y Vencerás → complejidad
3. Extensión a 2D
4. Extensión a 3D

## 1D - Fuerza bruta

La forma más directa de encontrar los dos más cercanos es comparar todas las parejas posibles y seleccionar la de menor distancia.

```python
def cercanos_1D_fuerza_bruta(puntos):
  n = len(puntos)
  mejor_distancia = float('inf')
  mejor_par = None

  for i in range(n):
    for j in range(i+1, n):
      distancia = abs(puntos[i] - puntos[j])
      if distancia < mejor_distancia:
        mejor_distancia = distancia
        mejor_par = (puntos[i], puntos[j])

  return mejor_par, mejor_distancia


puntos_1d = [3403, 4537, 9089, 9746, 7259, 1500, 2300, 9800, 4550]

par, distancia = cercanos_1D_fuerza_bruta(puntos_1d)
print("Par más cercano:", par)
print("Distancia:", distancia)
```

### Complejidad de la fuerza bruta (1D)

Comparar todas las parejas → O(n²).

¿Se puede mejorar? Sí: ordenando primero (O(n log n)), el par más cercano siempre son dos elementos consecutivos en el orden.

## 1D - Divide y Vencerás

Se ordena una vez. Como el par más cercano en una lista ordenada siempre son dos elementos consecutivos, no hace falta dividir en mitades ni recursividad: basta recorrer la lista ordenada comparando cada elemento con el siguiente.

```python
def cercanos_1D_divide_venceras(puntos):
  lista = sorted(puntos)

  mejor_distancia = float('inf')
  mejor_par = None

  for i in range(len(lista) - 1):
    distancia = abs(lista[i] - lista[i+1])
    if distancia < mejor_distancia:
      mejor_distancia = distancia
      mejor_par = (lista[i], lista[i+1])

  return mejor_par, mejor_distancia


par, distancia = cercanos_1D_divide_venceras(puntos_1d)
print("Par más cercano:", par)
print("Distancia:", distancia)
```

### Complejidad del Divide y Vencerás (1D)

Una vez ordenada la lista, el par más cercano siempre son dos elementos consecutivos en el orden: si no fueran adyacentes, cualquier punto entre ellos estaría más cerca de al menos uno de los dos. Por eso basta ordenar una vez y recorrer la lista comparando cada elemento con el siguiente.

Total: O(n log n) por el ordenamiento inicial, frente a O(n²) de la fuerza bruta.

## Extensión a 2D

Puntos (x, y), distancia euclídea. Fuerza bruta: igual, cambiando solo la distancia (O(n²)).

En 2D no hay un único orden que garantice vecinos consecutivos, así que el 1D no aplica directo. Mismo esquema de Divide y Vencerás:
1. Dividir por la mediana en x.
2. Vencer: recursión sobre cada mitad, obteniendo la mejor distancia entre las dos mitades.
3. Combinar: solo se revisan los puntos a menos de esa distancia de la línea divisoria (una franja), ordenados por y; cada punto se compara solo con los siguientes que estén a menos de esa distancia en y.

Cada llamada recursiva recibe los puntos ya ordenados por x y por y, sin reordenar en cada nivel.

```python
import math

def distancia(p, q):
  return math.hypot(p[0] - q[0], p[1] - q[1])


def cercanos_2D_fuerza_bruta(puntos):
  n = len(puntos)
  mejor_distancia = float('inf')
  mejor_par = None

  for i in range(n):
    for j in range(i+1, n):
      d = distancia(puntos[i], puntos[j])
      if d < mejor_distancia:
        mejor_distancia = d
        mejor_par = (puntos[i], puntos[j])

  return mejor_par, mejor_distancia


puntos_2d = [(1122, 6175), (135, 4076), (7296, 2741), (4123, 908), (9524, 3781), (3312, 7890), (1150, 6190), (8823, 120)]

par, dist = cercanos_2D_fuerza_bruta(puntos_2d)
print("Par más cercano:", par)
print("Distancia:", dist)
```

```python
def _cercanos_2D_dyv(px, py):
  n = len(px)

  if n <= 3: #caso base (fuerza bruta con pocos elementos)
    mejor_distancia = float('inf')
    mejor_par = None
    for i in range(n):
      for j in range(i+1, n):
        d = distancia(px[i], px[j])
        if d < mejor_distancia:
          mejor_distancia = d
          mejor_par = (px[i], px[j])
    return mejor_par, mejor_distancia

  #Dividir por la mediana en x
  medio = n // 2
  punto_medio = px[medio]

  px_izq = px[:medio]
  px_der = px[medio:]

  ids_izquierda = set(p[2] for p in px_izq) #ids para separar py sin volver a ordenar
  py_izq = [p for p in py if p[2] in ids_izquierda]
  py_der = [p for p in py if p[2] not in ids_izquierda]

  #Vencer (recursividad sobre cada mitad)
  par_izq, dist_izq = _cercanos_2D_dyv(px_izq, py_izq)
  par_der, dist_der = _cercanos_2D_dyv(px_der, py_der)

  if dist_izq <= dist_der:
    mejor_par, mejor_distancia = par_izq, dist_izq
  else:
    mejor_par, mejor_distancia = par_der, dist_der

  #Combinar: franja de ancho 2*mejor_distancia alrededor de la línea divisoria
  franja = [p for p in py if abs(p[0] - punto_medio[0]) < mejor_distancia]
  m = len(franja)
  for i in range(m):
    j = i + 1
    while j < m and (franja[j][1] - franja[i][1]) < mejor_distancia:
      d = distancia(franja[i], franja[j])
      if d < mejor_distancia:
        mejor_distancia = d
        mejor_par = (franja[i], franja[j])
      j += 1

  return mejor_par, mejor_distancia


def cercanos_2D_divide_venceras(puntos):
  #Se añade un id único a cada punto para poder repartir la lista ordenada por y
  #entre las dos mitades sin tener que reordenar en cada nivel de la recursión
  puntos_con_id = [(p[0], p[1], i) for i, p in enumerate(puntos)]
  px = sorted(puntos_con_id, key=lambda p: p[0])
  py = sorted(puntos_con_id, key=lambda p: p[1])

  (a, b), d = _cercanos_2D_dyv(px, py)
  return ((a[0], a[1]), (b[0], b[1])), d


par, dist = cercanos_2D_divide_venceras(puntos_2d)
print("Par más cercano:", par)
print("Distancia:", dist)
```

### Complejidad del Divide y Vencerás (2D)

El ordenamiento inicial (O(n log n)) no cambia el orden total.

Total: O(n log n), igual que en 1D y muy por debajo de O(n²).

## Extensión a 3D

Puntos (x, y, z). Fuerza bruta: solo cambia la distancia.

Divide y Vencerás: mismo esquema que en 2D (mediana en x, franja en la combinación), con un filtro extra por z antes de calcular la distancia completa.

Nota: en 2D el razonamiento de la franja se apoya en un solo criterio de orden (y). En 3D eso ya no es tan directo, así que aquí se usa una aproximación práctica (filtro extra por z), aunque menos ajustada en el peor caso que resolver la franja recursivamente con el algoritmo de una dimensión menos.

```python
def distancia_3D(p, q):
  return math.sqrt((p[0]-q[0])**2 + (p[1]-q[1])**2 + (p[2]-q[2])**2)


def cercanos_3D_fuerza_bruta(puntos):
  n = len(puntos)
  mejor_distancia = float('inf')
  mejor_par = None

  for i in range(n):
    for j in range(i+1, n):
      d = distancia_3D(puntos[i], puntos[j])
      if d < mejor_distancia:
        mejor_distancia = d
        mejor_par = (puntos[i], puntos[j])

  return mejor_par, mejor_distancia


puntos_3d = [(1122, 6175, 340), (135, 4076, 890), (7296, 2741, 1200), (4123, 908, 50), (9524, 3781, 4200), (1150, 6190, 355), (8823, 120, 60)]

par, dist = cercanos_3D_fuerza_bruta(puntos_3d)
print("Par más cercano:", par)
print("Distancia:", dist)
```

```python
def _cercanos_3D_dyv(px, py):
  n = len(px)

  if n <= 3: #caso base (fuerza bruta con pocos elementos)
    mejor_distancia = float('inf')
    mejor_par = None
    for i in range(n):
      for j in range(i+1, n):
        d = distancia_3D(px[i], px[j])
        if d < mejor_distancia:
          mejor_distancia = d
          mejor_par = (px[i], px[j])
    return mejor_par, mejor_distancia

  #Dividir por la mediana en x
  medio = n // 2
  punto_medio = px[medio]

  px_izq = px[:medio]
  px_der = px[medio:]

  ids_izquierda = set(p[3] for p in px_izq) #ids para separar py sin volver a ordenar
  py_izq = [p for p in py if p[3] in ids_izquierda]
  py_der = [p for p in py if p[3] not in ids_izquierda]

  #Vencer (recursividad sobre cada mitad)
  par_izq, dist_izq = _cercanos_3D_dyv(px_izq, py_izq)
  par_der, dist_der = _cercanos_3D_dyv(px_der, py_der)

  if dist_izq <= dist_der:
    mejor_par, mejor_distancia = par_izq, dist_izq
  else:
    mejor_par, mejor_distancia = par_der, dist_der

  #Combinar: franja en x, filtrando además por y y por z antes de medir distancia
  franja = [p for p in py if abs(p[0] - punto_medio[0]) < mejor_distancia]
  m = len(franja)
  for i in range(m):
    j = i + 1
    while j < m and (franja[j][1] - franja[i][1]) < mejor_distancia:
      if abs(franja[j][2] - franja[i][2]) < mejor_distancia:
        d = distancia_3D(franja[i], franja[j])
        if d < mejor_distancia:
          mejor_distancia = d
          mejor_par = (franja[i], franja[j])
      j += 1

  return mejor_par, mejor_distancia


def cercanos_3D_divide_venceras(puntos):
  #Se añade un id único a cada punto para poder repartir la lista ordenada por y
  #entre las dos mitades sin tener que reordenar en cada nivel de la recursión
  puntos_con_id = [(p[0], p[1], p[2], i) for i, p in enumerate(puntos)]
  px = sorted(puntos_con_id, key=lambda p: p[0])
  py = sorted(puntos_con_id, key=lambda p: p[1])

  (a, b), d = _cercanos_3D_dyv(px, py)
  return ((a[0], a[1], a[2]), (b[0], b[1], b[2])), d


par, dist = cercanos_3D_divide_venceras(puntos_3d)
print("Par más cercano:", par)
print("Distancia:", dist)
```

### Complejidad del Divide y Vencerás (3D)

O(n log n), asumiendo que la franja filtrada se mantiene lineal (cumplido en la práctica gracias a los filtros por y y z).

La cota formal ajustada al peor caso requiere resolver la franja recursivamente con el algoritmo de una dimensión menos. Aun así, O(n log n) sigue muy por debajo de O(n²).

## Datos aleatorios y comparativa de tiempos

Se generan puntos aleatorios de tamaño creciente y se mide el tiempo real: fuerza bruta vs. Divide y Vencerás, en 1D, 2D y 3D.

```python
import random
import time

def generar_puntos_1D(n, minimo=0, maximo=2_000_000):
  return random.sample(range(minimo, maximo), n)


def generar_puntos_2D(n, minimo=0, maximo=1_000_000):
  return [(random.randint(minimo, maximo), random.randint(minimo, maximo)) for _ in range(n)]


def generar_puntos_3D(n, minimo=0, maximo=1_000_000):
  return [(random.randint(minimo, maximo), random.randint(minimo, maximo), random.randint(minimo, maximo)) for _ in range(n)]


def medir_tiempo(funcion, *args):
  inicio = time.perf_counter()
  funcion(*args)
  return time.perf_counter() - inicio


tamanios = [100, 300, 600, 1000, 2000, 4000]
```

```python
tiempos_1D_fb = []
tiempos_1D_dv = []

print(f"{'n':>6} {'Fuerza bruta (s)':>18} {'Divide y Vencerás (s)':>24}")
for n in tamanios:
  puntos = generar_puntos_1D(n)
  t_fb = medir_tiempo(cercanos_1D_fuerza_bruta, puntos)
  t_dv = medir_tiempo(cercanos_1D_divide_venceras, puntos)
  tiempos_1D_fb.append(t_fb)
  tiempos_1D_dv.append(t_dv)
  print(f"{n:>6} {t_fb:>18.4f} {t_dv:>24.5f}")
```

```python
tiempos_2D_fb = []
tiempos_2D_dv = []

print(f"{'n':>6} {'Fuerza bruta (s)':>18} {'Divide y Vencerás (s)':>24}")
for n in tamanios:
  puntos = generar_puntos_2D(n)
  t_fb = medir_tiempo(cercanos_2D_fuerza_bruta, puntos)
  t_dv = medir_tiempo(cercanos_2D_divide_venceras, puntos)
  tiempos_2D_fb.append(t_fb)
  tiempos_2D_dv.append(t_dv)
  print(f"{n:>6} {t_fb:>18.4f} {t_dv:>24.5f}")
```

```python
tiempos_3D_fb = []
tiempos_3D_dv = []

print(f"{'n':>6} {'Fuerza bruta (s)':>18} {'Divide y Vencerás (s)':>24}")
for n in tamanios:
  puntos = generar_puntos_3D(n)
  t_fb = medir_tiempo(cercanos_3D_fuerza_bruta, puntos)
  t_dv = medir_tiempo(cercanos_3D_divide_venceras, puntos)
  tiempos_3D_fb.append(t_fb)
  tiempos_3D_dv.append(t_dv)
  print(f"{n:>6} {t_fb:>18.4f} {t_dv:>24.5f}")
```

```python
import matplotlib.pyplot as plt

fig, ejes = plt.subplots(1, 3, figsize=(15, 4))

datos = [
  ("1D", tiempos_1D_fb, tiempos_1D_dv),
  ("2D", tiempos_2D_fb, tiempos_2D_dv),
  ("3D", tiempos_3D_fb, tiempos_3D_dv),
]

for eje, (etiqueta, t_fb, t_dv) in zip(ejes, datos):
  eje.plot(tamanios, t_fb, marker='o', label='Fuerza bruta')
  eje.plot(tamanios, t_dv, marker='o', label='Divide y Vencerás')
  eje.set_title(etiqueta)
  eje.set_xlabel('n')
  eje.set_ylabel('tiempo (s)')
  eje.legend()

plt.tight_layout()
plt.show()
```

### Conclusión

Los tiempos confirman el análisis: fuerza bruta crece cuadrático, Divide y Vencerás crece n log n, en las tres dimensiones. La diferencia se dispara al aumentar n, y el patrón es el mismo en 1D, 2D y 3D.
