**Respuesta pregunta: ¿Cuántas posibilidades hay sin tener en cuenta las restricciones?**

Si no tenemos en cuenta ninguna restricción del enunciado, es decir, si pudiéramos repetir libremente una misma cifra o un mismo operador en distintos huecos, tenemos 9 opciones posibles para cada uno de los 5 huecos de cifra, y 4 opciones para cada uno de los 4 huecos de operador. Multiplicando todas esas opciones (9 por sí mismo 5 veces, y 4 por sí mismo 4 veces) obtenemos **9^5 × 4^4 = 15.116.544** combinaciones.

**Respuesta pregunta: ¿Cuántas posibilidades hay teniendo en cuenta todas las restricciones?**

Aplicando ahora las restricciones reales del enunciado (no podemos repetir una cifra ya usada, y los 4 operadores deben usarse todos exactamente una vez), el conteo cambia: para la primera cifra tenemos 9 opciones, para la segunda quedan 8 (ya no podemos repetir la primera), para la tercera 7, para la cuarta 6 y para la quinta 5. Multiplicando estas opciones obtenemos las formas de elegir y ordenar las 5 cifras: 9×8×7×6×5 = 15.120 (esto es la permutación P(9,5) = 9!/(9-5)!). Como los operadores no se repiten y deben usarse los 4, hay 4! = 24 formas de ordenarlos. Multiplicando ambos resultados: **15.120 × 24 = 362.880** posibilidades con restricciones.


````
# Para responder las dos preguntas separamos el conteo de las cifras del conteo
# de los operadores, y multiplicamos ambos resultados (regla del producto).

# --- Respuesta a: ¿posibilidades SIN restricciones? ---
# Si pudiéramos repetir libremente una misma cifra o un mismo operador en
# distintos huecos, tendríamos 9 opciones para cada uno de los 5 huecos de
# cifra, y 4 opciones para cada uno de los 4 huecos de operador.
sin_restricciones = 9**5 * 4**4
print(f"Sin restricciones: 9^5 (cifras) * 4^4 (operadores) = {sin_restricciones:,}")

# --- Respuesta a: ¿posibilidades CON restricciones? ---
# Aquí no quisimos confiar solo en la fórmula P(9,5)*4!: preferimos generar y
# contar las combinaciones directamente con itertools.permutations, que es el
# mismo enfoque que usamos más adelante para la fuerza bruta, así comprobamos
# que la fórmula y el conteo real coinciden.
combinaciones_cifras = list(itertools.permutations(range(1, 10), 5))          # 5 cifras distintas, en orden, elegidas de las 9 disponibles
combinaciones_operadores = list(itertools.permutations(['+', '-', '*', '/']))  # los 4 operadores, cada uno usado una sola vez

con_restricciones = len(combinaciones_cifras) * len(combinaciones_operadores)

print(f"Cifras: hay {len(combinaciones_cifras):,} formas de elegir y ordenar 5 de las 9 cifras (ej: {combinaciones_cifras[0]})")
print(f"Operadores: hay {len(combinaciones_operadores)} formas de ordenar los 4 operadores (ej: {combinaciones_operadores[0]})")
print(f"Con restricciones: {len(combinaciones_cifras):,} * {len(combinaciones_operadores)} = {con_restricciones:,}")

# Comprobamos que nuestro conteo directo coincide con la fórmula P(9,5)*4!...
formula = math.perm(9, 5) * math.factorial(4)
print(f"¿Coincide con la fórmula P(9,5)*4!? {con_restricciones == formula}")
# ...y, como notamos en la respuesta, también coincide con 9! (no es casualidad).
print(f"¿Coincide con 9!? 9! = {math.factorial(9):,} -> {con_restricciones == math.factorial(9)}")
````

> *Modelo para el espacio de soluciones*<br>(*) ¿Cual es la estructura de datos que mejor se adapta al problema? Argumentalo.(Es posible que hayas elegido una al principio y veas la necesidad de cambiar, arguentalo)

**Respuesta pregunta: ¿Cuál es la estructura de datos que mejor se adapta al problema?**

Para representar una solución candidata decidimos guardar dos cosas en un orden fijo: qué cifra ocupa cada uno de los 5 huecos, y qué operador ocupa cada uno de los 4 huecos, ya que el orden determina completamente el resultado de la expresión (no es lo mismo `4+2` que `2+4` si el resto de la expresión es distinto). La estructura que elegimos es una tupla de Python para cada una: `digitos = (d1,d2,d3,d4,d5)`, con 5 enteros distintos elegidos de {1,...,9}, y `operadores = (o1,o2,o3,o4)`, con una permutación de los símbolos `['+','-','*','/']`.

Optamos por tupla y no por lista porque, una vez que construimos una combinación candidata, no necesitamos modificarla, solo evaluarla y, si el resultado es un entero, guardarla en un conjunto (`set`) de resultados distintos. Una lista mutable no nos aportaba ninguna ventaja aquí y sí introducía el riesgo de modificarla por error mientras recorremos el espacio de soluciones. Tampoco necesitamos un diccionario, porque no buscamos por clave: solo recorremos las posiciones en orden, de la primera a la última.

````
# Decidimos representar cada solución candidata con dos tuplas. Usamos tupla y
# no lista porque, una vez que construimos una combinación, no la modificamos.
digitos = (4, 2, 6, 3, 1)          # 5 cifras distintas que elegimos de {1..9}, en el orden en que van en la expresión
operadores = ('+', '-', '/', '*')  # permutación de los 4 operadores; cada uno lo usamos una sola vez

def construir_expresion(digitos, operadores):
    """Intercalamos cifras y operadores, respetando el orden, para armar el string que luego evaluamos con eval()."""
    return f"{digitos[0]}{operadores[0]}{digitos[1]}{operadores[1]}{digitos[2]}{operadores[2]}{digitos[3]}{operadores[3]}{digitos[4]}"

expresion = construir_expresion(digitos, operadores)
print("Expresión construida:", expresion, "=", eval(expresion))

# Quisimos comprobar por qué nos importa guardar el ORDEN de la tupla y no solo
# qué cifras/operadores se usan: con las mismas cifras y los mismos operadores,
# pero reordenados, el resultado cambia por completo.
otro_orden = (1, 2, 3, 4, 6)
expresion2 = construir_expresion(otro_orden, operadores)
print("Mismas cifras, otro orden:", expresion2, "=", eval(expresion2))
print("¿Mismo resultado?", eval(expresion) == eval(expresion2))
````

**Respuesta pregunta: ¿Cuál es la función objetivo?**

Definimos como función objetivo la que transforma una solución candidata (la tupla de cifras y la tupla de operadores) en el número que queremos analizar: construimos el string de la expresión intercalando cifras y operadores, y lo evaluamos con la función `eval()` de Python. Formalmente: f(digitos, operadores) = eval("d1 o1 d2 o2 d3 o3 d4 o4 d5").

**Respuesta pregunta: ¿Es un problema de maximización o minimización?**

Consideramos que este problema no es exactamente un caso de maximización o minimización en el sentido clásico, como los que vimos en programación lineal, donde hay una función objetivo que se optimiza en una sola dirección. Aquí lo que se necesita es recorrer todo el espacio de soluciones y obtener el conjunto completo de resultados enteros posibles; una vez que tenemos ese conjunto, el máximo y el mínimo se obtienen directamente con las funciones max() y min(). Por eso, aunque el enunciado hable de un valor máximo y un valor mínimo, consideramos que el problema es, en el fondo, de enumeración exhaustiva. Esta es la razón principal por la que se descarta la técnica voraz: esta técnica toma una decisión en cada etapa sin poder reconsiderarla después, por lo que puede llegar a un valor alto o bajo, pero no garantiza que sea el máximo o el mínimo real, y tampoco permite saber si existen valores enteros que no se pueden alcanzar dentro del rango.

````
def f_objetivo(digitos, operadores):
    """
    Función objetivo: recibe la tupla de cifras y la tupla de operadores,
    arma la expresión intercalándolas en orden, y devuelve su valor numérico.
    Es la misma construcción que usamos en la Pregunta 2, la reutilizamos aquí
    como una función porque la vamos a llamar miles de veces en la búsqueda.
    """
    expr = f"{digitos[0]}{operadores[0]}{digitos[1]}{operadores[1]}{digitos[2]}{operadores[2]}{digitos[3]}{operadores[3]}{digitos[4]}"
    return eval(expr)

# Probamos la función con varias combinaciones para comprobar en la práctica que
# el resultado varía muchísimo según qué cifras/operadores elijamos: por eso el
# problema nos pide buscar tanto el máximo como el mínimo, no un único "mejor" valor.
print("Ejemplo del enunciado:", f_objetivo((4, 2, 6, 3, 1), ('+', '-', '/', '*')))
print("Otra combinación (da un valor alto):", f_objetivo((9, 8, 7, 6, 5), ('*', '+', '-', '/')))
print("Otra combinación (da un valor negativo):", f_objetivo((1, 9, 2, 8, 3), ('-', '/', '*', '+')))
````

**Respuesta pregunta: Diseña un algoritmo para resolver el problema por fuerza bruta**

Diseñamos el algoritmo de fuerza bruta generando absolutamente todas las combinaciones válidas de cifras y operadores, evaluándolas todas, y quedándonos con los resultados que fueran números enteros. Para generarlas usamos la función `itertools.permutations`: una vez para obtener las 24 posibles ordenaciones de los 4 operadores, y otra, para cada una de esas ordenaciones, para obtener las 15.120 posibles combinaciones ordenadas de 5 cifras elegidas de las 9 disponibles. Por cada combinación completa (cifras + operadores) construimos el string de la expresión y lo evaluamos con `eval()`; si el resultado era un número entero, lo agregábamos a un conjunto de resultados. Al terminar de recorrer las 362.880 combinaciones, obtuvimos el mínimo y el máximo del conjunto, y comparándolo contra el rango completo entre ambos, comprobamos si había huecos. Cabe mencionar que, al excluir el cero del conjunto de cifras disponibles, también se evita la posibilidad de una división entre cero al evaluar las expresiones con eval(), lo cual habría interrumpido la ejecución del algoritmo.

````
def fuerza_bruta(digitos_pool=range(1, 10), ops=('+', '-', '*', '/')):
    """
    Recorremos absolutamente todas las combinaciones válidas de cifras y
    operadores (sin descartar ninguna) y devolvemos el conjunto de resultados
    enteros distintos, junto con cuántas expresiones evaluamos en total.
    Dejamos digitos_pool y ops como parámetros (en vez de fijarlos adentro)
    para poder reutilizar esta misma función más adelante con otro conjunto
    de cifras, en la pregunta del dataset aleatorio.
    """
    resultados = set()   # usamos set y no lista porque solo nos interesan valores distintos
    evaluadas = 0
    for op_perm in itertools.permutations(ops):                    # las 24 formas de ordenar los operadores
        for dig_perm in itertools.permutations(digitos_pool, 5):   # las 15.120 formas de elegir y ordenar 5 cifras
            evaluadas += 1
            valor = f_objetivo(dig_perm, op_perm)
            if float(valor).is_integer():   # descartamos los resultados no enteros, como pide el enunciado
                resultados.add(int(valor))
    return resultados, evaluadas

# Medimos el tiempo con time.perf_counter() para poder comparar después contra
# la versión mejorada con poda (pregunta 6).
t_fb0 = time.perf_counter()
resultados_fb, evaluadas_fb = fuerza_bruta()
t_fb1 = time.perf_counter()

print(f"Expresiones evaluadas: {evaluadas_fb:,}")
print(f"Resultados enteros distintos: {len(resultados_fb)}")
print(f"Mínimo: {min(resultados_fb)}   Máximo: {max(resultados_fb)}")

# Para responder si hay "huecos", comparamos nuestro conjunto de resultados
# contra el rango completo de enteros entre el mínimo y el máximo que obtuvimos.
rango_completo = set(range(min(resultados_fb), max(resultados_fb) + 1))
faltantes = sorted(rango_completo - resultados_fb)
print(f"¿Se cubren todos los enteros del rango? {'Sí' if not faltantes else 'No'}  (faltantes: {len(faltantes)})")
print(f"Tiempo fuerza bruta: {t_fb1-t_fb0:.2f} s")
````

> Calcula la complejidad del algoritmo por fuerza bruta



**Respuesta pregunta: Calcula la complejidad del algoritmo por fuerza bruta**

Como cada expresión que evaluamos tiene una longitud fija (5 cifras y 4 operadores, siempre 9 símbolos), evaluarla con `eval()` cuesta un tiempo constante, O(1), independiente de cualquier otra variable. El costo total de nuestro algoritmo depende entonces únicamente de cuántas expresiones distintas generamos y evaluamos, que ya calculamos antes: P(9,5)×4! = 9! = 362.880. Por lo tanto, concluimos que la complejidad de nuestro algoritmo de fuerza bruta es **T(n) = O(n!)**, siendo n=9 el número de cifras disponibles en este problema concreto. Si generalizáramos a un número n de cifras disponibles (manteniendo fijos los 4 operadores), la cantidad de expresiones a evaluar sería O(P(n,5)·4!), que seguiría creciendo de forma factorial a medida que n aumenta.

````
# Queríamos ver qué tan rápido crece el problema si
# hubiera más cifras disponibles (aunque el enunciado fija n=9), para respaldar
# nuestra afirmación de que la complejidad es factorial y no, por ejemplo,
# exponencial o polinómica.
for n in [9, 12, 15, 18]:
    combinaciones = math.perm(n, 5) * math.factorial(4)   # misma fórmula P(n,5)*4! que dedujimos en la Pregunta 1
    print(f"n={n:2d} cifras disponibles -> {combinaciones:,} expresiones a evaluar")
````

**Respuesta pregunta: Diseña un algoritmo que mejore la complejidad del algoritmo por fuerza bruta. Argumenta por qué crees que mejora el algoritmo por fuerza bruta**

Mejoramos el algoritmo evitando evaluar expresiones que sabíamos de antemano que iban a dar el mismo resultado que otra ya evaluada. Esto ocurre cuando el operador que conecta dos cifras consecutivas es conmutativo (`+` o `*`), porque `a+b` da el mismo resultado que `b+a`, y lo mismo pasa con la multiplicación, sin importar el resto de la expresión que las rodea.

Aprovechando esa propiedad, construimos la solución con la técnica de vuelta atrás (backtracking): en vez de generar primero todas las permutaciones de cifras y evaluarlas después, las fuimos construyendo cifra a cifra, y en el momento de elegir la cifra que sigue a otra conectada por un operador conmutativo, exigimos que la nueva cifra fuera mayor que la anterior; si no lo era, descartábamos esa rama sin seguir explorándola (poda), porque su resultado ya estaba garantizado que iba a aparecer por la rama "espejo" donde el orden es el correcto.

Creemos que esta poda mejora el algoritmo porque no elimina ningún resultado posible, solo evita que volvamos a calcular uno que ya sabemos que es igual a otro, y comprobamos en la práctica que reduce las expresiones evaluadas de 362.880 a 75.600, un factor de reducción de aproximadamente **4,8 veces**, manteniendo exactamente el mismo mínimo, máximo y conjunto de enteros alcanzables que obtuvimos con la fuerza bruta.

Esto también coincide con lo que se explicó en la sesión de Vuelta Atrás: el diseño mejora cuando logramos identificar restricciones implícitas del problema, no solo las explícitas. La simetría de los operadores conmutativos es justamente eso, una restricción implícita que no está escrita en el enunciado, pero que pudimos deducir al analizar el problema.

````
# + y * son conmutativos (a+b=b+a, a*b=b*a); - y / no lo son. Cuando el
# operador entre dos cifras consecutivas es uno de estos dos, no nos hace
# falta explorar las dos formas de ordenarlas: con una alcanza.
COMMUTATIVOS = {'+', '*'}

def backtracking_podado(digitos_pool=list(range(1, 10)), ops=('+', '-', '*', '/')):
    """
    Misma idea que fuerza_bruta, pero en vez de generar primero todas las
    permutaciones de cifras con itertools y evaluarlas después, las armamos
    nosotros mismos cifra a cifra (backtracking) para poder cortar una rama
    apenas detectamos que es "espejo" de otra que ya vamos a explorar.
    """
    resultados = set()
    evaluadas = 0

    for op_perm in itertools.permutations(ops):
        usados = [False] * (max(digitos_pool) + 1)  # para no repetir cifra dentro de esta permutación de operadores
        secuencia = [0] * 5                           # aquí vamos guardando la cifra elegida en cada posición

        def rec(pos):
            nonlocal evaluadas
            if pos == 5:
                # Llegamos a una expresión completa: la evaluamos igual que en fuerza_bruta
                evaluadas += 1
                valor = f_objetivo(tuple(secuencia), op_perm)
                if float(valor).is_integer():
                    resultados.add(int(valor))
                return
            for d in digitos_pool:
                if usados[d]:
                    continue
                # PODA: si el operador que conecta la cifra anterior con esta es
                # conmutativo, solo seguimos si la nueva cifra es mayor que la
                # anterior — así evitamos explorar la rama "espejo" (que ya
                # cubre el mismo resultado desde el otro orden).
                if pos > 0 and op_perm[pos-1] in COMMUTATIVOS and secuencia[pos-1] > d:
                    continue
                usados[d] = True
                secuencia[pos] = d
                rec(pos + 1)     # seguimos construyendo la expresión con esta cifra puesta
                usados[d] = False  # y la liberamos al volver, para probar otra cifra en su lugar

        rec(0)

    return resultados, evaluadas

# Medimos el tiempo igual que hicimos con fuerza_bruta, para poder comparar ambas versiones.
t_pod0 = time.perf_counter()
resultados_mejor, evaluadas_mejor = backtracking_podado()
t_pod1 = time.perf_counter()

print(f"Expresiones evaluadas (con poda): {evaluadas_mejor:,}")
print(f"Factor de reducción vs fuerza bruta: {evaluadas_fb/evaluadas_mejor:.2f}x")
print(f"Resultados enteros distintos: {len(resultados_mejor)}")
print(f"Mínimo: {min(resultados_mejor)}   Máximo: {max(resultados_mejor)}")
# Este chequeo es clave: si la poda estuviera mal hecha y hubiera descartado
# algún resultado válido, esta comparación nos lo mostraría como False.
print(f"¿Mismos resultados que fuerza bruta? {resultados_mejor == resultados_fb}")
print(f"Tiempo backtracking podado: {t_pod1-t_pod0:.2f} s")
````

> (*)Calcula la complejidad del algoritmo

**Respuesta pregunta: Calcula la complejidad del algoritmo**

La poda que aplicamos reduce el número de expresiones que efectivamente evaluamos, pero no cambia el peor caso teórico del algoritmo: en el peor caso posible (si, hipotéticamente, ningún operador de la permutación fuera conmutativo) no descartaríamos ninguna rama y el algoritmo se comportaría exactamente igual que la fuerza bruta. Por eso concluimos que, en el **peor caso sigue siendo O(n!)**, la misma clase que la fuerza bruta, porque la poda solo quita un factor **constante** (como mucho ~2× por cada operador conmutativo presente, y hay como mucho 2 de los 4 operadores que son conmutativos: `+` y `*`).

Lo que sí cambia, y pudimos medirlo, es la constante oculta detrás de esa notación: empíricamente, nuestro algoritmo con poda evalúa **4,8 veces menos** expresiones que la fuerza bruta para resolver exactamente el mismo problema. Preferimos ser honestos aquí y no confundir esto con una mejora de clase de complejidad (por ejemplo, pasar de factorial a polinómico): eso no ocurre en nuestro caso, es una mejora real de rendimiento, pero de constante, no de clase.

Y en esa misma sesión, se menciona que la vuelta atrás suele tener complejidad O(2^n), pero esa cota es para problemas de selección de subconjuntos, como la mochila que vimos. Nuestro problema es distinto, porque genera permutaciones, así que la complejidad que le corresponde es O(n!), como ya calculamos.
